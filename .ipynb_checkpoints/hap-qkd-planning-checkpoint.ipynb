```python
from libraries import *
```

```python
gp.setParam("OutputFlag", 1)
```

```python
gnodes, hnodes, links, demands = init_setup_real_planning()
```

```python
hap_list = range(1, len(hnodes))

key_bottle_list_kmeans    = []
key_bottle_list_greedy    = []
key_bottle_list_heuristic = []

key_avg_list_kmeans    = []
key_avg_list_greedy    = []
key_avg_list_heuristic = []
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])
    
    solution, keyrate_bottle, keyrate_avg = placement_kmeans(gnodes, hnodes_list_iter, links)

    key_bottle_list_kmeans.append(keyrate_bottle)
    key_avg_list_kmeans.append(keyrate_avg)
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])
    
    solution, keyrate_bottle, keyrate_avg = placement_greedy(gnodes, hnodes_list_iter, links)

    key_bottle_list_greedy.append(keyrate_bottle)
    key_avg_list_greedy.append(keyrate_avg)
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])

    links = []
    # Links: connect only GSs to HAPs
    for gs_node in gnodes:
        for hap_node in hnodes_list_iter:
            links.append(link(gs_node, hap_node,
                              [100]*len(syst.T),
                              [(0,0,0)]*len(syst.T),
                              [1e6]*len(syst.T)))
            links.append(link(hap_node, gs_node,
                              [100]*len(syst.T),
                              [(0,0,0)]*len(syst.T),
                              [1e6]*len(syst.T)))
    
    solution, keyrate_bottle, keyrate_avg = placement(gnodes, hnodes_list_iter, links)

    key_bottle_list_heuristic.append(keyrate_bottle)
    key_avg_list_heuristic.append(keyrate_avg)
```

```python
print(f"key_bottle_list_kmeans: {key_bottle_list_kmeans}")
print(f"key_bottle_list_greedy: {key_bottle_list_greedy}")
print(f"key_bottle_list_heuristic: {key_bottle_list_heuristic}")

print(f"key_avg_list_kmeans: {key_avg_list_kmeans}")
print(f"key_avg_list_greedy: {key_avg_list_greedy}")
print(f"key_avg_list_heuristic: {key_avg_list_heuristic}")

plt.figure(figsize=(4,3))
plt.plot(demand_list, key_bottle_list_kmeans, label="K-means", linestyle="-", marker="o", markevery=2)
plt.plot(demand_list, key_bottle_list_greedy, label="Greedy", linestyle="-", marker="^", markevery=2)
plt.plot(hap_list, key_bottle_list_heuristic, label="MIP", linestyle="-", marker="x", markevery=2)

plt.xlabel("# HAPs")
plt.ylabel("Bottleneck E2E SKR (bps)")
plt.yscale("log")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.show()

plt.figure(figsize=(4,3))
plt.plot(hap_list, key_avg_list_kmeans, label="K-means", linestyle="-", marker="o", markevery=2)
plt.plot(hap_list, key_avg_list_greedy, label="Greedy", linestyle="-", marker="^", markevery=2)
plt.plot(hap_list, key_avg_list_heuristic, label="MIP", linestyle="-", marker="x", markevery=2)

plt.xlabel("# HAPs")
plt.ylabel("Average E2E SKR (bps)")
plt.yscale("log")
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```
