```python
gnodes, hnodes, links, demands = init_setup_real_planning()
lon0 = 279; lat0 = 49
```

```python
hap_list = range(1, len(hnodes))

key_bottle_list_kmeans        = []
key_bottle_list_greedy        = []
key_bottle_list_heuristic_btl = []
key_bottle_list_heuristic_avg = []
key_bottle_list_upper         = []

key_avg_list_kmeans        = []
key_avg_list_greedy        = []
key_avg_list_heuristic_btl = []
key_avg_list_heuristic_avg = []
key_avg_list_upper         = []
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])
    
    solution, keyrate_bottle, keyrate_avg = placement_kmeans(gnodes, hnodes_list_iter, links)

    key_bottle_list_kmeans.append(keyrate_bottle)
    key_avg_list_kmeans.append(keyrate_avg)
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])
    
    solution, keyrate_bottle, keyrate_avg = placement_greedy(gnodes, hnodes_list_iter, links)

    key_bottle_list_greedy.append(keyrate_bottle)
    key_avg_list_greedy.append(keyrate_avg)
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])

    links = []
    # Links: connect only GSs to HAPs
    for gs_node in gnodes:
        for hap_node in hnodes_list_iter:
            links.append(link(gs_node, hap_node,
                              [100]*len(syst.T),
                              [(0,0,0)]*len(syst.T),
                              [1e6]*len(syst.T)))
            links.append(link(hap_node, gs_node,
                              [100]*len(syst.T),
                              [(0,0,0)]*len(syst.T),
                              [1e6]*len(syst.T)))

    #print(links)
    
    #solution, keyrate_bottle, keyrate_avg = placement_path(gnodes, hnodes_list_iter, links, "btl")
    solution, keyrate_bottle, keyrate_avg = placement_path_lower(gnodes, hnodes_list_iter, links, "btl")

    key_bottle_list_heuristic_btl.append(keyrate_bottle)
    key_avg_list_heuristic_btl.append(keyrate_avg)
```

```python
hnodes_list_iter = []

for h_num in hap_list:
    hnodes_list_iter.append(hnodes[h_num-1])

    links = []
    # Links: connect only GSs to HAPs
    for gs_node in gnodes:
        for hap_node in hnodes_list_iter:
            links.append(link(gs_node, hap_node,
                              [100]*len(syst.T),
                              [(0,0,0)]*len(syst.T),
                              [1e6]*len(syst.T)))
            links.append(link(hap_node, gs_node,
                              [100]*len(syst.T),
                              [(0,0,0)]*len(syst.T),
                              [1e6]*len(syst.T)))

    #print(links)
    
    #solution, keyrate_bottle, keyrate_avg = placement_path(gnodes, hnodes_list_iter, links, "avg")
    solution, keyrate_bottle, keyrate_avg = placement_path_lower(gnodes, hnodes_list_iter, links, "avg")

    key_bottle_list_heuristic_avg.append(keyrate_bottle)
    key_avg_list_heuristic_avg.append(keyrate_avg)
```

```python
dist_bottle = 0.0
dist_sum    = 0.0
num_pairs   = 0

mean_alt = sum(hnodes[0].H) / len(hnodes[0].H)

# GS pairs
for i, gs1 in enumerate(gnodes):
    gs1_c1, gs1_c2 = latlon_to_tangent(gs1.lg, gs1.la, lon0, lat0)

    for j, gs2 in enumerate(gnodes):
        if j <= i:
            continue

        gs2_c1, gs2_c2 = latlon_to_tangent(gs2.lg, gs2.la, lon0, lat0)
        hap_c1, hap_c2 = (gs1_c1+gs2_c1) / 2, (gs1_c2+gs2_c2) / 2

        # Best HAP for this pair
        pair_best = math.inf

        dist = math.sqrt((gs1_c1 - hap_c1)**2 +
                        (gs1_c2 - hap_c2)**2 +
                        mean_alt**2)

        dist_sum    += dist
        dist_bottle = max(dist_bottle, dist)
        num_pairs  += 1

print(f"dist_bottle: {dist_bottle}, dist_avg: {dist_sum/num_pairs}")

# Compute efficiencies
eta_theory = ts.channel_theory(direction="downlink", gs_alt=0, hap_alt=sum(hnodes[0].H) / len(hnodes[0].H),
                               distance=dist_bottle, n_correction=6, params_in=ts.params)
# Compute SKRs
keyrate_bottle = ts.compute_skr(eta_theory)

# Compute efficiencies
eta_theory = ts.channel_theory(direction="downlink", gs_alt=0, hap_alt=sum(hnodes[0].H) / len(hnodes[0].H),
                               distance=dist_sum / num_pairs, n_correction=6, params_in=ts.params)
# Compute SKRs
keyrate_avg = ts.compute_skr(eta_theory)

for h_num in hap_list:
    key_bottle_list_upper.append(keyrate_bottle)
    key_avg_list_upper.append(keyrate_avg)
```
